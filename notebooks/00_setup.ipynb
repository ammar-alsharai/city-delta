```sql
-- City Lens: one-time setup of the catalog, schemas and raw volume
CREATE CATALOG IF NOT EXISTS city_delta
COMMENT 'City Lens: daily pipeline of public city data';

CREATE SCHEMA IF NOT EXISTS city_delta.bronze
COMMENT 'Raw data as received from each source';

CREATE SCHEMA IF NOT EXISTS city_delta.silver
COMMENT 'Parsed and cleaned data';

CREATE SCHEMA IF NOT EXISTS city_delta.gold
COMMENT 'Joined tables for the dashboard';

CREATE VOLUME IF NOT EXISTS city_delta.bronze.raw
COMMENT 'Raw source files (XML, JSON, gzip)';
```
